```python
import numpy as np
import matplotlib.pyplot as plt

x = [1, 2, 3, 5, 6, 7, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 88, 78, 62, 55, 50, 52, 58, 65, 70, 76, 82, 85, 88, 94, 100, 105, 108]

# Create polynomial model
degree = 2
coefficients = np.polyfit(x, y, degree)
polynomial = np.poly1d(coefficients)

# Trend line
x_line = np.linspace(min(x), max(x), 100)
y_line = polynomial(x_line)

# Predict speed at 17:00
time = 17
predicted_speed = polynomial(time)

print("Predicted speed at 17:00:", predicted_speed)

# Plot data
plt.scatter(x, y, label="Actual data")
plt.plot(x_line, y_line, label="Polynomial trend")

# Predicted point
plt.scatter(
    time,
    predicted_speed,
    color="red",
    s=100,
    label=f"Prediction at 17:00 = {predicted_speed:.2f}"
)

# Show predicted value
plt.annotate(
    f"{predicted_speed:.2f}",
    (time, predicted_speed),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.xlabel("Time")
plt.ylabel("Speed")
plt.title("Vehicle Speed Prediction\nАбдурахманов Азизихон")

plt.legend()
plt.grid()
plt.show()
```
